```python


import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="DEATH_EVENT",
    hue="DEATH_EVENT",
    palette="Set2",
    legend=False
)

plt.title("Distribution of Death Events")
plt.xlabel("Death Event")
plt.ylabel("Number of Patients")
plt.xticks([0, 1], ["No Death Event", "Death Event"])
plt.tight_layout()
plt.show()
```

We noticed `Death_event` is a categorical variable: Yes or No. We will replace/convert it to an integer as the ML model requires numeric variables. 

```python
df['DEATH_EVENT'].unique()
```

```python
df['DEATH_EVENT'] = df['DEATH_EVENT'].replace({"Yes":1, "No":0})
```
