```python

features = df.drop(columns=["DEATH_EVENT", "time"])
target = df["DEATH_EVENT"]

pearson = features.corrwith(target, method="pearson")
spearman = features.corrwith(target, method="spearman")

corr_comparison = pd.DataFrame({
    "Pearson": pearson,
    "Spearman": spearman
})

corr_comparison["Absolute_Pearson"] = (
    corr_comparison["Pearson"].abs()
)

corr_comparison = corr_comparison.sort_values(
    "Absolute_Pearson",
    ascending=False
)

display(corr_comparison)
```

```python

plot_data = corr_comparison[
    ["Pearson", "Spearman"]
].sort_values("Pearson")

plot_data.plot(
    kind="barh",
    figsize=(10, 7)
)

plt.title("Feature Correlations with Death Event")
plt.xlabel("Correlation Coefficient")
plt.ylabel("Clinical Features")
plt.axvline(0, color="black", linewidth=0.8)
plt.tight_layout()
plt.show()
```

```python

features = [
    "age",
    "ejection_fraction",
    "serum_creatinine",
    "serum_sodium"
]

fig, axes = plt.subplots(2, 2, figsize=(13, 9))

for ax, feature in zip(axes.flat, features):
    sns.boxplot(
        data=df,
        x="DEATH_EVENT",
        y=feature,
        hue="DEATH_EVENT",
        palette="Set2",
        legend=False,
        ax=ax
    )

    ax.set_title(f"{feature} by Outcome")
    ax.set_xlabel("Death Event")

plt.tight_layout()
plt.show()
```

```python

summary = df.groupby("DEATH_EVENT")[
    [
        "age",
        "ejection_fraction",
        "serum_creatinine",
        "serum_sodium"
    ]
].agg(["mean", "median", "std"])

display(summary.round(2))
```

```python
corr_pearson = df.corr(method='pearson')['DEATH_EVENT'].sort_values(key=abs, ascending=False)[1:].head(10)
corr_pearson
```

For both methods, we notice weak or moderate levels of correlation between Churn and a given variable. 
* Ideally, we pursue strong correlation levels. However, this is not always possible.

We will consider the top five correlation levels at `df_ohe` and will study the associated variables at `df`

```python

# Pearson correlation
corr_pearson = (
    df.corr(method='pearson')['DEATH_EVENT']
    .drop('DEATH_EVENT')
    .sort_values(key=abs, ascending=False)
    .head(10)
)

# Spearman correlation
corr_spearman = (
    df.corr(method='spearman')['DEATH_EVENT']
    .drop('DEATH_EVENT')
    .sort_values(key=abs, ascending=False)
    .head(10)
)

# Select top 5 features from each method
top_n = 5

selected_features = set(
    corr_pearson.head(top_n).index.to_list()
    + corr_spearman.head(top_n).index.to_list()
)

print("Selected features:", selected_features)
print("Number of selected features:", len(selected_features))
```

## Variables Distribution by Churn

We plot the distribution (numerical and categorical) coloured by Churn

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')


def plot_categorical(df, col, target_var):

    plt.figure(figsize=(12, 5))
    sns.countplot(data=df, x=col, hue=target_var, order=df[col].value_counts().index)
    plt.xticks(rotation=90)
    plt.title(f"{col}", fontsize=20, y=1.05)
    plt.show()


def plot_numerical(df, col, target_var):
    plt.figure(figsize=(8, 5))
    sns.histplot(data=df, x=col, hue=target_var, kde=True, element="step")
    plt.title(f"{col}", fontsize=20, y=1.05)
    plt.show()


target_var = 'DEATH_EVENT'
for col in vars_to_study:
    if df_eda[col].dtype == 'object':
        plot_categorical(df_eda, col, target_var)
        print("\n\n")
    else:
        plot_numerical(df_eda, col, target_var)
        print("\n\n")
```

```python

from feature_engine.discretisation import ArbitraryDiscretiser

# Define age boundaries
age_bins = {
    "age": [0, 50, 65, 80, float("inf")]
}

# Create discretiser
discretiser = ArbitraryDiscretiser(
    binning_dict=age_bins,
    return_object=False
)

# Preserve the original data
df_discrete = discretiser.fit_transform(df.copy())

# Compare original and discretised age
display(
    pd.DataFrame({
        "Original_Age": df["age"],
        "Age_Group": df_discrete["age"]
    }).head(10)
)
```

```python

from feature_engine.discretisation import ArbitraryDiscretiser
import numpy as np

# Define age intervals
age_map = [-np.inf, 50, 60, 70, 80, np.inf]

# Create discretiser
disc = ArbitraryDiscretiser(
    binning_dict={'age': age_map}
)

# Apply discretisation
df_parallel = disc.fit_transform(df.copy())

# Preview
display(df_parallel.head())
```

```python

comparison = pd.DataFrame({
    'Original_Age': df['age'],
    'Discretised_Age': df_parallel['age']
})

display(comparison.head(15))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df_parallel,
    x='age',
    hue='DEATH_EVENT',
    palette='Set2'
)

plt.title('Age Categories by Death Event')
plt.xlabel('Age Category')
plt.ylabel('Number of Patients')
plt.legend(title='Death Event')
plt.tight_layout()
plt.show()
```
